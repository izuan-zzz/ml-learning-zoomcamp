```python
asia = df[df['origin'] == 'Asia']
```

```python
data = asia[['vehicle_weight', 'model_year']].head(7)
data
```

```python
import numpy as np
X = data.values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
print(w, w.sum(),3))
```
